## 2. Let's practice using some dataframe methods.

Find the mean and standard deviation of the weather data.

````python
avg = frame.agg({
    'Temp' : 'mean',
    'Rainfall' : 'mean',
})
avg
````

````python
stdv = frame.agg({
        'Temp' : 'std',
    'Rainfall' : 'std',
})
stdv
````
